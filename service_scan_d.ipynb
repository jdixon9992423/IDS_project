```python
import pandas as pd
import matplotlib.pyplot as plt
import os

here = os.path.abspath('') #get the current directory the python file is in, just place all teh other files in this folder

service_scan_csv1_file= os.path.join(here, 'IOT_service_scan_2711_n.csv')
#exfiltration_csv2_file= os.path.join(here, 'exfiltration_2.csv')

#os_scan_csv=os.path.join(here, 'OS_Scan_labels.csv')

df=pd.read_csv(service_scan_csv1_file)
#df2=pd.read_csv(exfiltration_csv2_file)
#df3=pd.read_csv(os_scan_csv,sep=";")

```

```python
protocol_vals={'UDP': 5,'TCP':4,'0x0806':3,'':2,'0x8035':1,'IPv4':0, '0x86dd':6}
df['Protocol']=[protocol_vals[k] for k in df['Protocol']]
```

```python


df['Window']=df['Window'].fillna(0)
df['dport']=df['dport'].fillna(0)
df['sport']=df['sport'].fillna(0)
df['TCP_Flags']=df['TCP_Flags'].fillna('0x000')
df['Time to Live']=df['Time to Live'].fillna(0)
df['Bytes in flight']=df['Bytes in flight'].fillna(0)
df['Time since previous frame(UDP)']=df['Time since previous frame(UDP)'].fillna(0)
df['Time since previous frame in this TCP stream(TCP)']=df['Time since previous frame in this TCP stream(TCP)'].fillna(0)

```

```python
from ast import literal_eval
df['TCP_Flags']=[literal_eval(k) for k in df['TCP_Flags']]
```

```python
X=df[['sport','dport','Protocol','Length','Window','TCP_Flags','Bytes in flight','Time since previous frame in this TCP stream(TCP)','Time since previous frame(UDP)']].values
Y=df['Attack'].values
```

```python
X.shape
```

```python
#X=sc.fit_transform(X)
```

```python
X
```

```python
import sklearn
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()

#X=sc.fit_transform(X)

scaler=sc.fit(X)
X=scaler.transform(X)

import pickle
pickle.dump(sc, open('service_scan_scaler_.pkl','wb'))#should i put the scaler instead of sc???
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, Y_train, Y_test= train_test_split(X,Y,test_size=0.25)
```

```python
X_train
```

```python
from sklearn import svm
from sklearn.ensemble import AdaBoostClassifier 
from sklearn import tree
from sklearn.neural_network import MLPClassifier
```

```python
svm_model=svm.SVC()
adaboost_model=AdaBoostClassifier()
decision_tree=tree.DecisionTreeClassifier()
mlp_model=MLPClassifier(hidden_layer_sizes=(10,7,4))
```

```python
svm_model.fit(X_train,Y_train)
```

```python
adaboost_model.fit(X_train,Y_train)
```

```python
decision_tree.fit(X_train,Y_train)
```

```python
#mlp_model=MLPClassifier(hidden_layer_sizes=(10,8,4))
mlp_model.fit(X_train,Y_train)
```

```python
svm_model.score(X_test,Y_test)
```

```python
adaboost_model.score(X_test,Y_test)
```

```python
decision_tree.score(X_test,Y_test)
```

```python
mlp_model.score(X_test,Y_test)
```
