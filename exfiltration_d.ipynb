```python


df['Window']=df['Window'].fillna(0)
df['dport']=df['dport'].fillna(0)
df['sport']=df['sport'].fillna(0)
df['TCP_Flags']=df['TCP_Flags'].fillna('0x000')
df['Bytes in flight']=df['Bytes in flight'].fillna(0)
df['Bytes sent since last PSH flag']=df['Bytes sent since last PSH flag'].fillna(0)
df['The RTT to ACK the segment was']=df['The RTT to ACK the segment was'].fillna(0)
df['Time since previous frame in this TCP stream(TCP)']=df['Time since previous frame in this TCP stream(TCP)'].fillna(0)
#df['Time to Live']=df['Time to Live'].fillna(0)


```

```python
X=df[['sport','dport','Protocol','Length','Window','TCP_Flags','Bytes in flight','Bytes sent since last PSH flag','Time since previous frame in this TCP stream(TCP)']].values
Y=df['Attack'].values
```

```python
import sklearn
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()

#X=sc.fit_transform(X)

scaler=sc.fit(X)
X=scaler.transform(X)

import pickle
pickle.dump(sc, open('exfiltration_scaler_.pkl','wb'))
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, Y_train, Y_test= train_test_split(X,Y,test_size=0.25)
```

```python
X_train
```

```python
from sklearn import svm
from sklearn.ensemble import AdaBoostClassifier 
from sklearn.neural_network import MLPClassifier
from sklearn import tree
```

```python
svm_model=svm.SVC()
adaboost_model=AdaBoostClassifier()
mlp_model=MLPClassifier(hidden_layer_sizes=(10,7,4))
decision_tree=tree.DecisionTreeClassifier()
```

```python
svm_model.fit(X_train,Y_train)  #i am hereeeeeeeeeeeeeeeee dec 06
```

```python
adaboost_model.fit(X_train,Y_train)
```

```python
#mlp_model=MLPClassifier(hidden_layer_sizes=(10,7,4))
mlp_model.fit(X_train,Y_train)
```

```python
decision_tree.fit(X_train,Y_train)
```

```python
svm_model.score(X_test,Y_test)
```

```python
adaboost_model.score(X_test,Y_test)
```

```python
mlp_model.score(X_test,Y_test)
```

```python
decision_tree.score(X_test,Y_test)
```
